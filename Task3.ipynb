# Task 3: Data Visualization Dashboard

## Introduction

In this task, data visualizations are created using Python, Pandas, and Matplotlib to present customer sales data in a clear and understandable way.

The dashboard-style visualizations will help identify sales patterns, product performance, customer distribution, payment preferences, and daily sales trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_customer_sales_dataset.csv")
```

```python
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
product_sales = df.groupby("Product")["Total_Amount"].sum()

plt.figure(figsize=(8, 5))

plt.bar(product_sales.index, product_sales.values)

plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")

plt.tight_layout()
plt.show()
```

```python
df["Product"] = df["Product"].astype("string").str.strip().str.title()
```

```python
product_sales = df.groupby("Product")["Total_Amount"].sum()

print(product_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(product_sales.index, product_sales.values)

plt.title("Product-wise Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")

plt.tight_layout()
plt.show()
```

```python
daily_sales = df.groupby("Purchase_Date")["Total_Amount"].sum()

plt.figure(figsize=(10, 5))

plt.plot(daily_sales.index, daily_sales.values, marker="o")

plt.title("Daily Sales Trend")
plt.xlabel("Purchase Date")
plt.ylabel("Total Sales (₹)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Payment_Method"] = (
    df["Payment_Method"]
    .astype("string")
    .str.strip()
    .replace({
        "upi": "UPI",
        "credit card": "Credit Card",
        "Cash on Delivery": "COD"
    })
)
```

```python
print(df["Payment_Method"].value_counts())
```

```python
payment_count = df["Payment_Method"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    payment_count.values,
    labels=payment_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Method Distribution")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=8, edgecolor="black")

plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Quantity"], df["Total_Amount"])

plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales (₹)")

plt.tight_layout()
plt.show()
```
